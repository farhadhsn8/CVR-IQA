```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load the three images
img1 = np.array(Image.open('I77.png'))
img2 = np.array(Image.open('I77_24_05.png'))
img3 = np.array(Image.open('I77_09_05.png'))

# Function to crop random patches
def crop_random_patches(images, num_patches, patch_size):
    h, w = images[0].shape[:2]
    patches = []
    for _ in range(num_patches):
        top = np.random.randint(0, h - patch_size)
        left = np.random.randint(0, w - patch_size)
        patches.append([img[top:top+patch_size, left:left+patch_size] for img in images])
    return patches

# Crop 10 random patches of size 224x224 from each image
patches = crop_random_patches([img1, img2, img3], num_patches=10, patch_size=224)

# Create a 5x2 subplot for each image
fig, axes = plt.subplots(5, 6, figsize=(20, 15))

for i, image_patches in enumerate(zip(*patches)):
    for j, patch in enumerate(image_patches):
        row = j // 2
        col = (j % 2) + i * 2
        axes[row, col].imshow(patch)
        axes[row, col].axis('off')

plt.tight_layout()
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load the three images
img1 = np.array(Image.open('0240.png'))
img2 = np.array(Image.open('0240.png'))
img3 = np.array(Image.open('0240.png'))

# Function to crop random patches
def crop_random_patches(images, num_patches, patch_size):
    # Resize images to 512 width while maintaining aspect ratio
    resized_images = []
    for img in images:
        aspect_ratio = img.shape[1] / img.shape[0]
        new_height = int(512 / aspect_ratio)
        resized_img = np.array(Image.fromarray(img).resize((512, new_height), Image.LANCZOS))
        resized_images.append(resized_img)
    images = resized_images
    h, w = images[0].shape[:2]
    patches = []
    for _ in range(num_patches):
        top = np.random.randint(0, h - patch_size)
        left = np.random.randint(0, w - patch_size)
        patches.append([img[top:top+patch_size, left:left+patch_size] for img in images])
    return patches

# Crop 10 random patches of size 224x224 from each image
patches = crop_random_patches([img1, img2, img3], num_patches=10, patch_size=224)

# Create a 5x2 subplot for each image
fig, axes = plt.subplots(5, 6, figsize=(20, 15))

for i, image_patches in enumerate(zip(*patches)):
    for j, patch in enumerate(image_patches):
        row = j // 2
        col = (j % 2) + i * 2
        axes[row, col].imshow(patch)
        axes[row, col].axis('off')

plt.tight_layout()
plt.show()

```
